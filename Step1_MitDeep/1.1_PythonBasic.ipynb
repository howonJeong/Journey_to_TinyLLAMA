브로드캐스트 - 곱하는 차원이 달라도 element wise 연산이 가능하도록 복제

```python
#브로드캐스트
a = np.array([[1,2],[2,1]])
b = np.array([10,20])
print(a*b)
```

```python
a = np.array([[1,2], [2,1]])
b = np.array(2)
print(a*b)
```

```python
b = np.array([1,2,3])
print(a*b)
```

```python
a = np.array([[1,2],[3,4]])
for row in a:
    print(row)
for column in a:
    print(column)
```

```python
print(a[0]) #0행(0th row)
print(a[1])
print(a[:, 1:2])
```

```python
A = a.flatten()
print(A)
print(A[np.array([0,2])])
#인덱스를 배열로 전달하여 여러 원소를 얻을 수 있다.

#특정 조건을 만족하는 원소만 얻기가 가능
print(A < 3)
print(A[A<3]) #오오...
```

매트플롯

```python
import matplotlib.pyplot as plt
```

```python
x = np.arange(0, 6, 0.4) 
y = np.sin(x)
y_cos = np.cos(x)

plt.plot(x, y, label="sin")
plt.plot(x, y_cos, linestyle="--", label="cos")

plt.xlabel("x")
plt.ylabel("y")
plt.title('sin & cos')
plt.legend()
plt.show()
```
