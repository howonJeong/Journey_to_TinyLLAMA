```python
def AND(x1, x2):
    w1, w2, theta = 0.5, 0.5, 0.7
    ans = x1*w1 + x2*w2
    if  ans <= theta:
        return 0
    else :
        return 1
```

```python
print([AND(1, 0), AND(1,1)])
```

```python
def AdvAND(x1, x2):
    x = np.array([x1, x2])
    w = np.array([0.5, 0.5])
    b = 0.7
    ans = np.sum(x*w) - b
    if ans < 0:
        return 0
    else:
        return 1
```

```python
print([AdvAND(1, 1), AdvAND(0, 1)])
```

```python
def NAND(x1, x2):
    x = np.array([x1, x2])
    w = np.array([0.5, 0.5])
    b = 0.7
    ans = np.sum(x*w) - b
    if ans < 0:
        return 1
    else:
        return 0
```

```python
def OR(x1, x2):
    x = np.array([x1, x2])
    w = np.array([0.5, 0.5])
    b = 0.4
    if np.sum(x*w)-b < 0:
        return 0
    else:
        return 1
    
print(OR(1,1), OR(0,1), OR(0,0))
```

AND NAND OR은 가중치 매개변수의 차이일 뿐이지 같은 구조의 퍼셉트론이다!

```python
def XOR(x1, x2):
    x = np.array([x1, x2])
    w = np.array([0.5, 0.5])
    b = 0.5
    if np.sum(x*w)-b == 0:
        return 1
    else:
        return 0
    
print(XOR(1,1), XOR(0,1), XOR(0,0))
```

```python
print(XOR(1,0))
```

```python
def RealXOR(x1, x2):
    return AND(NAND(x1,x2), OR(x1,x2))

print(f"된다고 얘는 {RealXOR(0, 0.99)}")
```
